```python
from datetime import datetime

# Список газет с их датами
newspapers = [
    {"name": "The Moscow Times", "date_str": "Wednesday, October 2, 2002"},
    {"name": "The Guardian", "date_str": "Friday, 11.10.13"},
    {"name": "Daily News", "date_str": "Thursday, 18 August 1977"}
]

# Определяем форматы для каждой газеты
date_formats = {
    "The Moscow Times": "%A, %B %d, %Y",
    "The Guardian": "%A, %d.%m.%y",
    "Daily News": "%A, %d %B %Y"
}

# Переводим каждую дату в объект datetime и выводим результат
for paper in newspapers:
    name = paper["name"]
    date_str = paper["date_str"]

    try:
        dt_object = datetime.strptime(date_str, date_formats[name])
        print(f"{name}: {dt_object}")
    except ValueError as e:
        print(f"{name}: Ошибка парсинга - {e}")
```

```python
from datetime import datetime, timedelta

def date_range(start_date, end_date):
    try:
        # Пробуем преобразовать строки в объекты datetime
        start = datetime.strptime(start_date, "%Y-%m-%d")
        end = datetime.strptime(end_date, "%Y-%m-%d")
        
        # Проверяем, что начальная дата не больше конечной
        if start > end:
            print(f"Ошибка: начальная дата ({start_date}) больше конечной ({end_date})")
            return []
        
        # Генерируем список дат
        result = []
        current = start
        
        while current <= end:
            # Добавляем дату в формате YYYY-MM-DD
            result.append(current.strftime("%Y-%m-%d"))
            # Переходим к следующему дню
            current += timedelta(days=1)
        
        return result
        
    except ValueError as e:
        # Обрабатываем ошибки некорректного формата или несуществующей даты
        print(f"Ошибка: {e}")
        return []


print("\nТест 1: date_range('2022-01-01', '2022-01-03')")
result1 = date_range('2022-01-01', '2022-01-03')
print(f"Результат: {result1}")

print("\nТест 2: date_range('2022-01-03', '2022-01-01')")
result2 = date_range('2022-01-03', '2022-01-01')
print(f"Результат: {result2}")

print("\nТест 3: date_range('2022-02-30', '2022-02-31')")
result3 = date_range('2022-02-30', '2022-02-31')
print(f"Результат: {result3}")

print("\nТест 4: date_range('2022/01/01', '2022-01-03')")
result4 = date_range('2022/01/01', '2022-01-03')
print(f"Результат: {result4}")

print("\nТест 5: date_range('2022-01-28', '2022-02-02')")
result5 = date_range('2022-01-28', '2022-02-02')
print(f"Результат: {result5}")

print("\nПримеры работы с объектами datetime:")

now = datetime.now()
print(f"Текущая дата и время: {now}")
print(f"Только дата: {now.date()}")
print(f"Только время: {now.time()}")

formatted_date = now.strftime("%d.%m.%Y %H:%M:%S")
print(f"Отформатированная дата: {formatted_date}")

tomorrow = now + timedelta(days=1)
yesterday = now - timedelta(days=1)
print(f"Завтра: {tomorrow.date()}")
print(f"Вчера: {yesterday.date()}")

date1 = datetime(2023, 1, 1)
date2 = datetime(2023, 12, 31)
diff = date2 - date1
print(f"Дней между 01.01.2023 и 31.12.2023: {diff.days}")
```
