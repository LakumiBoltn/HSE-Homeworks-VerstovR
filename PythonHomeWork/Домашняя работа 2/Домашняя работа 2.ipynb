```python
def sum_distance(from_num, to_num):
    a, b = sorted([from_num, to_num])
    return (b - a + 1) * (a + b) // 2

print("Проваерка")

result1 = sum_distance(8, 12)
print(f"sum_distance(8, 12) = {result1}")

result2 = sum_distance(12, 8)
print(f"sum_distance(12, 8) = {result2}")

result3 = sum_distance(4, 4)
print(f"sum_distance(4, 4) = {result3}")

result4 = sum_distance(-1, 1)
print(f"sum_distance(-1, 1) = {result4}")
```

```python
def trim_and_repeat(text, offset=0, repetitions=1):
    return text[offset:] * repetitions

print("Проверка")

result1 = trim_and_repeat("hello")
print(f"1. trim_and_repeat('hello') = '{result1}'")

result2 = trim_and_repeat("hello", 2)
print(f"2. trim_and_repeat('hello', 2) = '{result2}'")

result3 = trim_and_repeat("hello", 0, 3)
print(f"3. trim_and_repeat('hello', 0, 3) = '{result3}'")

result4 = trim_and_repeat("hello", 2, 2)
print(f"4. trim_and_repeat('hello', 2, 2) = '{result4}'")

result5 = trim_and_repeat("abc", 3, 5)
print(f"5. trim_and_repeat('abc', 3, 5) = '{result5}'")
```
