```python
def middle_letters(word):
    length = len(word)
    if length == 0:
        return ""

    middle = length // 2
    return word[middle] if length % 2 == 1 else word[middle - 1:middle + 1]

test_cases = ['test', 'testing', 'a', 'ab', 'hello']
for word in test_cases:
    result = middle_letters(word)
    print(f"word = '{word}' Результат: '{result}'")
```

 – среднюю букву, если число букв в слове нечётное;

 Пример работы программы:

```python
def datings(boys, girls):
    if len(boys) != len(girls):
        return "Кто-то останется без пары"

    sorted_boys = sorted(boys)
    sorted_girls = sorted(girls)

    pairs = []
    for i in range(len(sorted_boys)):
        pairs.append(f"{sorted_boys[i]} и {sorted_girls[i]}")

    result = "Идеальные пары:\n" + "\n".join(pairs)
    return result


print("Тест 1:")
boys1 = ['Oscar', 'Alex', 'Aleksei', 'Mikhail', 'Evgeniy']
girls1 = ['Yana', 'Evgeniya', 'Stefaniya', 'Olesya', 'Anfisa']
print(datings(boys1, girls1))

print("\nТест 2:")
boys2 = ['Alexey', 'Borislav', 'Prokhor', 'Nikolai', 'Borislav', 'Michael']
girls2 = ['Varya', 'Valentina', 'Kira', 'Venera', 'Liliya']
print(datings(boys2, girls2))

print("\nДополнительные тесты:")

print("Тест 3 (пустой список):")
boys3 = []
girls3 = ['Vasilisa']
print(datings(boys3, girls3))

print("\nТест 4 (по одному человеку):")
boys4 = ['Mikhail']
girls4 = ['Anna']
print(datings(boys4, girls4))

print("\nТест 5 (девушек больше):")
boys5 = ['Tom', 'Boris']
girls5 = ['Evdokiya', 'Marya', 'Svetlana']
print(datings(boys5, girls5))
```
