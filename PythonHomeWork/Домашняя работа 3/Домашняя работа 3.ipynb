* Программа перевода содержимого файла в словарь

```python
import json
import os

# Проверяем существование папки Data и файла
if not os.path.exists('Data'):
    print("Ошибка: папка 'Data' не найдена в текущей директории")
    print("Текущая директория:", os.getcwd())
else:
    print("Папка Data найдена")
    
    # Путь к файлу
    file_path = 'Data/purchase_log.txt'
    
    if not os.path.exists(file_path):
        print(f"Ошибка: файл {file_path} не найден")
    else:
        print(f"Файл {file_path} найден")
        
        # Чтение файла и создание словаря purchases
        purchases = {}
        
        with open(file_path, 'r', encoding='utf-8') as file:
            for line_num, line in enumerate(file, 1):
                line = line.strip()
                if line:  # Пропускаем пустые строки
                    try:
                        record = json.loads(line)
                        user_id = record.get('user_id')
                        category = record.get('category')
                        if user_id and category:
                            purchases[user_id] = category
                    except json.JSONDecodeError:
                        print(f"Предупреждение: строка {line_num} не является корректным JSON")
        
        print(f"\nВсего записей в словаре purchases: {len(purchases)}")
        
        # Вывод первых двух элементов словаря
        print("\nПервые два элемента словаря purchases:")
        count = 0
        for user_id, category in purchases.items():
            if count < 2:
                print(f"{user_id} '{category}'")
                count += 1
            else:
                break
```

* Программа склеивания файлов

```python
import csv
import json
import os

# Задание 1 (повторяем для использования в задании 2)
purchases = {}
with open('purchase_log.txt', 'r', encoding='utf-8') as file:
    for line in file:
        record = json.loads(line.strip())
        purchases[record['user_id']] = record['category']

print(f"Словарь purchases содержит {len(purchases)} записей")
print("Первые 2 записи из purchases:")
count = 0
for user_id, category in purchases.items():
    if count < 2:
        print(f"{user_id} '{category}'")
        count += 1
    else:
        break
print()

# Задание 2 - создание funnel.csv
# Удаляем старый файл funnel.csv, если он существует
if os.path.exists('funnel.csv'):
    os.remove('funnel.csv')
    print("Старый файл funnel.csv удален")

# Построчная обработка visit_log.csv и запись в funnel.csv
with open('visit_log.csv', 'r', encoding='utf-8') as visit_file, \
     open('funnel.csv', 'w', encoding='utf-8', newline='') as funnel_file:
    
    # Создаем reader и writer
    visit_reader = csv.DictReader(visit_file)
    funnel_writer = csv.writer(funnel_file)
    
    # Записываем заголовок в funnel.csv
    funnel_writer.writerow(['user_id', 'source', 'category'])
    
    # Обрабатываем каждую строку visit_log.csv
    rows_written = 0
    for row in visit_reader:
        user_id = row['user_id']
        
        # Проверяем, есть ли покупка для данного user_id
        if user_id in purchases:
            category = purchases[user_id]
            funnel_writer.writerow([user_id, row['source'], category])
            rows_written += 1

print(f"В файл funnel.csv записано {rows_written} строк")

# Вывод первых трех строк из созданного файла funnel.csv
print("\nПервые 3 строки файла funnel.csv:")
with open('funnel.csv', 'r', encoding='utf-8') as funnel_file:
    funnel_reader = csv.reader(funnel_file)
    for i, row in enumerate(funnel_reader):
        if i < 4:  # Выводим заголовок + 3 строки данных
            print(', '.join(row))
        else:
            break
```
